```python
#check outlier
sns.boxplot(data= df)
plt.show()
```

```python
#univariant(Analysis of a single variable)
sns.histplot(df['area'])
plt.show()
```

```python
#bivariant analysis
sns.scatterplot(x=df['male_life_expectancy'], y =df['female_life_expectancy'])
plt.show()
```

```python
#Numerical
sns.histplot(df['birth_rate'])
plt.show()
```

```python
#categorical
sns.countplot(x=df['capital'])
plt.xticks(rotation=45)
plt.show()
sns.countplot(x=df['region'])
plt.xticks(rotation=45)
plt.show()
```

```python
corr = df.select_dtypes(include=['number']).corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Europe Dataset Correlation Heatmap")
plt.show()
```
